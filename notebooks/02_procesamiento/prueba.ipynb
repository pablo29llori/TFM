```python
def euro_to_float(x):
    if pd.isna(x):
        return pd.NA
    x = str(x).replace('€', '').replace(',', '').strip()
    return float(x) if x else pd.NA
```

```python
for col in ['gross_weekly_eur', 'gross_yearly_eur', 'adj_gross_eur']:
    df_sal[col] = df_sal[col].apply(euro_to_float)

df_sal['age'] = pd.to_numeric(df_sal['age'], errors='coerce')
```

```python
import unicodedata
import re
```

```python
def normalize_text(s):
    if pd.isna(s):
        return pd.NA
    s = str(s)
    s = unicodedata.normalize('NFKD', s).encode('ascii', 'ignore').decode('ascii')
    s = s.lower().strip()
    s = re.sub(r'[^a-z0-9\s]', ' ', s)
    s = re.sub(r'\s+', ' ', s)
    return s
```

```python
df_stats['player_norm'] = df_stats['player'].apply(normalize_text)
df_stats['team_norm'] = df_stats['team'].apply(normalize_text)

df_sal['player_norm'] = df_sal['player'].apply(normalize_text)
df_sal['team_norm'] = df_sal['team'].apply(normalize_text)
```

```python
stats_teams = sorted(df_stats['team'].dropna().unique().tolist())
sal_teams = sorted(df_sal['team'].dropna().unique().tolist())

print("Equipos en stats:")
print(stats_teams)

print("\nEquipos en salarios:")
print(sal_teams)
```

```python
team_map = {
    'Alaves': 'Deportivo Alavés',
    'Atletico Madrid': 'Atlético Madrid',
    'Girona': 'Girona FC',
    'Leganes': 'Leganés',
    'Valladolid': 'Real Valladolid'
}
```

```python
df_sal['team'] = df_sal['team'].replace(team_map)
df_sal['team_norm'] = df_sal['team'].apply(normalize_text)
```

```python
stats_teams = sorted(df_stats['team'].dropna().unique().tolist())
sal_teams = sorted(df_sal['team'].dropna().unique().tolist())

print("Equipos en stats:")
print(stats_teams)

print("\nEquipos en salarios:")
print(sal_teams)
```

```python
df_final = df_stats.merge(
    df_sal[['player_norm', 'team_norm', 'gross_weekly_eur', 'gross_yearly_eur']],
    on=['player_norm', 'team_norm'],
    how='left'
)
```

```python
matched = df_final['gross_yearly_eur'].notna().sum()
total = len(df_final)

print(f"Matches: {matched}/{total}")
print(f"Tasa de match: {matched/total:.2%}")
```

```python
unmatched = df_final[df_final['gross_yearly_eur'].isna()].copy()
unmatched[['player', 'team']].drop_duplicates().sort_values(['team', 'player']).head(50)
```

```python
from difflib import SequenceMatcher

salary_names_by_team = (
    df_sal.groupby('team_norm')['player_norm']
    .apply(list)
    .to_dict()
)

rows = []

for _, row in unmatched[['player', 'team', 'player_norm', 'team_norm']].drop_duplicates().iterrows():
    candidates = salary_names_by_team.get(row['team_norm'], [])
    
    best_match = None
    best_score = 0
    
    for cand in candidates:
        score = SequenceMatcher(None, row['player_norm'], cand).ratio()
        if score > best_score:
            best_score = score
            best_match = cand
    
    rows.append({
        'player_stats': row['player'],
        'team': row['team'],
        'best_salary_match_norm': best_match,
        'score': best_score
    })

possible_matches = pd.DataFrame(rows).sort_values('score', ascending=False)
possible_matches.head(50)
```
